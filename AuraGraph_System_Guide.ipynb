# AuraGraph – Complete System & User Navigation Guide

## Overview
AuraGraph is an AI-powered study platform that helps students create personalized study materials from PDFs, slides, and textbooks, then test their knowledge through adaptive quizzes and visual knowledge graphs.

---

## 1️⃣ AUTHENTICATION LAYER

### Entry Point: Login Page

#### Three Authentication Options:

**A) Sign In (Existing Users)**
- Email address
- Password
- Redirects to Dashboard on success

**B) Register (New Users)**
- Name (auto-filled from email if blank)
- Email address
- Password (minimum 8 characters required)
- Creates account and redirects to Dashboard

**C) Try Demo**
- No login required
- One-click access to pre-built DSP notebook
- Uses local storage (offline mode)
- Perfect for testing the platform
- Can switch to login/register anytime

### Session Management:
- Session token stored in `localStorage`
- Auto-redirect to Dashboard after login
- Logout clears all session data

## 5️⃣ QUIZ SYSTEM (Assessment Hub)

### Access Quiz Center:
- **Button:** "Quiz Center" in sidebar
- **Keyboard Shortcut:** Ctrl/Cmd + Shift + Q

### Quiz Types:

#### 1. **General Test** (10 Questions)
- Mixed difficulty questions
- Covers all concepts
- Comprehensive assessment

**How to Take:**
1. Dashboard → Quiz Center
2. Click **Take General Test**
3. Answer each question (A/B/C/D or 1/2/3/4)
4. View results with breakdown

#### 2. **Focus Test** (5 Questions) – Weak Areas
- Targets struggling concepts
- Only available if weak concepts exist
- Focused practice on problem areas

**How to Take:**
1. Dashboard → Quiz Center
2. Click **Take Sniper Test**
3. System auto-selects weak concepts
4. Answer focused questions
5. Review and retry or move on

#### 3. **Concept-Wise Test** (5 Questions) ✅ NEW
- From ONE selected concept
- User manually selects concept
- Targeted deep-dive learning

**How to Take:**
1. Dashboard → Quiz Center
2. Click **Take Concept Quiz**
3. Select a concept from grid (shows status: ✅/⚠️/🎯/📚)
4. Click **Start Concept Quiz**
5. Answer 5 questions
6. View results

### During Quiz:
- **Answer Options:** Press A-D or 1-4 keys
- **Next Question:** Press Enter or click button
- **Instant Feedback:** See correct/incorrect after each question
- **Explanation:** View detailed explanation

### After Quiz (Results Screen):
- **Score Display:** X/5 (or X/10 for General)
- **Percentage & Color:** Green (70%+), Orange (40-69%), Red (<40%)
- **XP Earned:** With accuracy bonus if 80%+
- **Feedback Message:**
  - ≥70%: "Excellent concept clarity"
  - 40-69%: "Nice progress — revise weak sub-parts"
  - <40%: "Let's spend more time on this concept"

### Post-Quiz Actions:

**If Score < 70%:**
- **🔄 Retry This Concept** → Retake same quiz
- **📖 Revise This Concept** → Jump to that section in notes
- **Back** → Return to Quiz Center

**If Score ≥ 70%:**
- **📖 Revise This Concept** (optional)
- **Done** → Close and celebrate! 🎉

### Quiz History & Filtering:

**View Quiz History:**
1. Dashboard → Quiz Center
2. Click **Concept History** tab
3. See all attempted concept quizzes

**Each Attempt Shows:**
- Concept name (e.g., "Concept Attempt 1 • History")
- Score (e.g., "4/5 (80%)")
- Timestamp
- **View review** link

**Filter by Concept:**
1. Open Concept History
2. Use dropdown: "Filter by concept"
3. Select concept name
4. See only those attempts

**Detailed Review:**
1. Click "View review" on any attempt
2. See question-by-question breakdown
3. Your answer vs correct answer
4. Full explanation per question
5. Click "Back to History" to exit

### Concept Status Types:

1. **🟢 Mastered**

```
1. Login / Dashboard
2. Click "New Notebook"
3. Enter notebook title
4. Upload PDF/Image/PPTX/DOCX
5. Click "Generate Notes"
6. AI creates study notes
7. Read & annotate content
8. Track mastery in Study Hub
```

## 1️⃣2️⃣ AI ASSISTANT PROMPT TEMPLATE

### System Context for AI Assistant:

You are an intelligent assistant for AuraGraph, an AI-powered study platform. Your role is to help users navigate and use the platform efficiently.

**Core Instructions:**
1. Always provide step-by-step instructions
2. Reference exact features/modules (Dashboard, Notebook, Study Hub, Quiz, etc.)
3. Prefer actionable navigation steps over lengthy explanations
4. If multiple ways exist to accomplish a task, suggest the simplest path first
5. Keep answers concise but clear
6. Include optional tips if they improve the experience

**Response Format:**
```
## How to [Task]

### Step-by-Step:
1. [Action] → [Navigation/Button name]
2. [Action] → [Result]
3. [Action] → [Result]
4. [Action] → [Outcome]

### 💡 Optional Tips:
- [Shortcut or helpful tip]
- [Alternative method if relevant]
```

**User Question:** "How do I take a concept quiz?"

**Assistant Response:**
```
## How to Take a Concept-Wise Quiz

### Step-by-Step:
1. **Open Quiz Center** → Click Quiz Center button in sidebar (or press Ctrl+Shift+Q)
2. **Click "Take Concept Quiz"** → Opens concept selection modal
3. **Select a concept** → Choose from grid (shows status: ✅/⚠️/🎯/📚)
4. **Click "Start Concept Quiz"** → Begin 5-question quiz
5. **Answer questions** → Press A/B/C/D or 1/2/3/4 keys
6. **Review results** → See score, XP earned, and feedback

### 💡 Optional Tips:
- **Keyboard shortcut:** Press Ctrl/Cmd + Shift + Q to open Quiz Center directly
- **Low scores?** Click "📖 Revise This Concept" to jump straight to that section in notes
- **Retry feature:** If score < 70%, you can immediately retake the same quiz
```

## 1️⃣5️⃣ TROUBLESHOOTING & HELP

### Common Issues & Solutions:

**1. Can't upload file**
- Check file format (PDF, PNG, JPG, PPTX, DOCX only)
- Ensure file size < 50MB
- Try refreshing page and re-uploading

**2. Notes not generating**
- Ensure content was extracted successfully
- Try adding custom prompt for clarity
- Click "Regenerate" for alternative version

**3. Quiz not loading**
- Check internet connection
- Clear browser cache
- Refresh page and try again

**4. Annotations not saving**
- Ensure you're not in read-only mode
- Try creating annotation again
- Refresh page to see saved annotations

**5. Can't find concept in graph**
- Use search box to find concept
- Check spelling/capitalization
- Ensure concept exists in notebook

### Getting Help:
1. Click **Feedback Widget** (bottom-right)
2. Describe issue in detail
3. Optionally attach screenshot
4. Submit → Team reviews and responds
